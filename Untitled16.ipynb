# Exercise 5 — **uint8**

```python
arr = np.array([0,100,200,255])
```

```python
arr = np.array([0,100,200,255], dtype=np.uint8)
result = arr + 20
print(result)
```

```python
arr = np.array([0,100,200,255],dtype=np.uint8)
arr16 = arr.astype(np.int16)
result= arr16 + 20
result = np.clip(result, 0, 255)

result = result.astype(np.uint8)

print(result)
```

```python
import cv2
import numpy as np

# TODO 1: Load an image in grayscale
gray = cv2.imread("/content/Marvel (2).jpg", cv2.IMREAD_GRAYSCALE)
```

# Load the color image

```python

color = cv2.imread("/content/Marvel (2).jpg", cv2.IMREAD_COLOR)
```

# TODO 3: Split the color image

```python


B, G, R = cv2.split(color)
```

# TODO 4: Merge the channels

```python

reconstructed = cv2.merge([B, G, R])
```

# TODO 5: Verify reconstruction

```python

print("Reconstruction successful:",
      np.array_equal(color, reconstructed))
```
